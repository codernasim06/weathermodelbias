# Data Exploration Notebook

This notebook inspects the IEEE COMPAS '26 project structure and reads the repository files to provide a practical starting point for data exploration and modeling.

```python
from pathlib import Path
import os
import json

# Locate the repo root from the notebook location
notebook_dir = Path.cwd()
if notebook_dir.name == 'notebooks':
    repo_root = notebook_dir.parent
else:
    repo_root = notebook_dir

repo_root
```

```python
# List project files and folders
for path in sorted(repo_root.rglob('*')):
    if path.is_file() and path.suffix.lower() not in {'.png', '.jpg', '.jpeg', '.gif', '.svg', '.pdf'}:
        print(path.relative_to(repo_root))
```

```python
# Read and preview the project documentation files
docs_to_read = ['README.md', 'ARCHITECTURE.md', 'requirements.txt']

for filename in docs_to_read:
    file_path = repo_root / filename
    if file_path.exists():
        print(f'\n===== {filename} =====')
        print(file_path.read_text(encoding='utf-8'))
```

```python
# Read the reusable Python modules in src/
src_files = sorted((repo_root / 'src').glob('*.py'))
for src_file in src_files:
    print(f'\n===== {src_file.relative_to(repo_root)} =====')
    print(src_file.read_text(encoding='utf-8'))
```

```python
# Inspect the data folders and check whether the raw input file exists
raw_dir = repo_root / 'data' / 'raw'
processed_dir = repo_root / 'data' / 'processed'

print('Raw data files:')
for path in sorted(raw_dir.iterdir()):
    print(' -', path.name)

print('\nProcessed data files:')
for path in sorted(processed_dir.iterdir()):
    print(' -', path.name)
```
